```python
#Prompt the user to input a temperature in the format ( e.g. , 45F, 102C, etc.)
tem = input ("input the temperature you like to convert (e.g., 45F, 102c, etc.):")

# extract the numerical part of the temperature and convert it to an integer
degree = int(tem[:-1])

# extract the convention part of the temperature input (either'C' OR 'F')
i_convention = tem[-1]

#check if the input convention is in uppercase 'C'(Celsive)
if i_convention.upper()=="C":
    #convert the celsius temperature to Fahrenheit
    result = int(round((9* degree) / + 32))
    o_convention = "Fahrenheit" # set the output convention as Fahrenheit
# check (f the input convention is in uppercase 'F' (Fahrenheit)
elif i_convention.upper() == "F":
    # convert the Fahrenhiet temperature to Celsius
    result = int(round((degree-32)*5/9))
    o_convention = "Celsius"  # set the output convention as Celius
else:
    # if the input convention is neither'C'nor'F',print an error message and exit the program
    print("input proper convention.")
    quit()
# display the converted tenmperature in the specified output convention 
print("the temperature in ",o_convention,"is",result,"degree.")
```

```python
# function to multiply two numbers
def multiply(a,b):
    result = a * b
    return result

    # function to divide two numbers
def divide(a,b):
    if b != 0:
        result = a / b
    else:
        result = "cannot divide by zero"
    return result

    # main program
num1 = 10
num2 = 5

# calling the function and printing the results
print("addition:", add(num1, num2))
print("subtraction:", subtract(num1, num2))
print("multiplication :", multiply(num1, num2))
print("Division:", divide(num1,num2))
```

```python
# Positional arguments

def greet(name,age):
    print(f"hello {name}, you are {age} years old.")

greet("ramar", 30)    
    
```

```python
# keyword arguments

def greet(name,age):
    print(f"hello {name}, you are {age} years old.")

greet (age=30, name = "maria") # keyword arguments    
```

```python
#default arguments
def greet(name, age=25):
    print(f"hello {name} , you are {age} years old .")

greet("ramar") #uses default age value
greet("maria", 30) #overrides default age value
```

```python
#variable-lenth arguments

def print_numbers(*args):
    for number in args:
        print(number)
print_numbers(1,2,3,4,5) # variable - lenth positional arguments        
        
```

```python
# keyword variable-lenth arguments(**kwargs)

def print_info(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="kumar", professison = "engineer")
print_info(country= "india", population=1.4e9 , capital="new delhi")
```

```python
# mixing different argument types
def mixed_arguments(a, b=2, *args, **kwargs):
    print(f"a:{a}")
    print(f"b:{b}")
    print("args:",args)
    print("kwargs:",kwargs)

mixed_arguments(1,3,4,5, x=10, y=20)    

    
```
